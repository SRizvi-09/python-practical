```python
# import librarires
import pandas as pd
```

```python
# load the data in games.csv as data frame called games
games = pd.read_csv('games.csv')
```

```python
#reassign games as its subset of the columns:
# Games_date, Games_status_text, team_id_home, team_id_away, points_home, points_away, home_team_wins
# use double brackets [[ ]]

games = games[['GAME_DATE_EST',
               'GAME_STATUS_TEXT',
               'TEAM_ID_home',
               'TEAM_ID_away',
               'PTS_home',
               'PTS_away',
               'HOME_TEAM_WINS']]
```

```python
#look at the new games dataframe's first 5 rows and info summ
games.head()
```

```python
# convert game_date to a datetime dtype
games['GAME_DATE_EST'] = pd.to_datetime(games['GAME_DATE_EST'])
```

```python
games['GAME_DATE_EST'].dtype
```

```python
# convert game_status_text to a string dtype
games['GAME_STATUS_TEXT'] = games['GAME_STATUS_TEXT'].astype('string')
```

```python
games['GAME_STATUS_TEXT'].dtype
```
